### Pathlib

why to use Path?
<br> the code will work across operating systems
<br> automatically uses correct sepperator \ / 
<br> the code is easier to manipulate

```python
from pathlib import Path
```

```python
f = open(Path(r'D:\_Python\pokus2.txt'), "r") # this is one way to do it. Ideally I should close the file, so that other programs can access it.
f.read()
f.close()
```

```python
folder = Path(r'D:\_Python')
file = folder / "pokus2.txt"
f = open(file, "r")
```

```python
# but if something happens in between, the file might still not be accessible to other programs, so a better way is to use with:
with open(r'D:\_Python\pokus2.txt', "r") as f: #it will automatically call the close statement when the execution leaves the block
    print(f.read())
```

## Pandas


### Excel

```python
import pandas as pd

sheet1 = pd.DataFrame({'a': [1, 2, 3], 'b': [4, 5, 6]})
sheet2 = pd.DataFrame({'a': [1, 2, 3], 'b': [4, 5, 6]})

file_name_xls = 'example.xlsx'
sheet1.to_excel(file_name_xls, sheet_name='Example1', )
```

```python
# more options with ExcelWriter
with pd.ExcelWriter(file_name_xls) as writer:
    sheet1.to_excel(writer, sheet_name='Sheet1')
    sheet2.to_excel(writer, sheet_name='Sheet2')
    
pd.read_excel(file_name_xls, sheet_name='Sheet1', index_col=0, header= None)
```

### csv

```python
file_name_csv = 'example_csv.csv'
sheet1.to_csv(file_name_csv)
pd.read_csv(file_name_csv)

```

```python
sheet1.to_csv(file_name_csv, index=False, sep= ',')
pd.read_csv(file_name_csv)
```

## File Manipulations

<br> "r" - Read - Default value. Opens a file for reading, error if the file does not exist
<br> "a" - Append - Opens a file for appending, creates the file if it does not exist
<br> "w" - Write - Opens a file for writing, creates the file if it does not exist
<br> "x" - Create - Creates the specified file, returns an error if the file exists

### Read

```python
with open(Path(r'D:\_Python\pokus2.txt'), "r") as f:
    print(f.read())
```

```python
# specify the number of characters to be read
with open(Path(r'D:\_Python\pokus2.txt'), "r") as f:
    print(f.read(5))
```

```python
# or read the entire first line
with open(Path(r'D:\_Python\pokus2.txt'), "r") as f:
    print(f.readline())
```

```python
# it iterates row over row
with open(Path(r'D:\_Python\pokus2.txt'), "r") as f:
    print(f.readline())
    print(f.readline())
```

```python
# to go through the whole file line by line
for i in open(Path(r'D:\_Python\pokus2.txt'), "r"):
    print(i)
```

```python
f.close() # make sure to close the file so that other processes can access it
```

### Amend

```python
# a just appends the file
with open(Path(r'D:\_Python\pokus2.txt'), "a") as f:
    f.write('just added a line\n')
    f.close() # not neccessary here because I used "with"

with open(Path(r'D:\_Python\pokus2.txt'), "r") as f: print(f.read())
```

### Write

```python
# w rewrites the entire file
with open(Path(r'D:\_Python\pokus2.txt'), "w") as f:
    f.write('first line\n')
    f.write('second line\n')
    f.close() # not neccessary here because I used "with"

with open(Path(r'D:\_Python\pokus2.txt'), "r") as f: print(f.read())
```

### Create a file - X

```python
new = open('created_file.txt', 'x')
new.close()

with open(Path(os.getcwd()) / "created_file.txt", "w") as f:
    f.write('first line')
    
with open(Path(os.getcwd()) / "created_file.txt", "r") as f:
    print(f.read())
```

```python
# removes a file
os.remove('created_file.txt')
```
