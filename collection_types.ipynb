## Collection/Container Types

- <b>Sequence Types: </b> list, tuple, range
- <b>Mapping Type:</b> dict
- <b>Set Types:</b>  set, frozenset


-  List
   - ordered
   - changeable
   - allows duplicate members
-  Tuple
    - ordered
    - unchangeable (but lists and other types that are changeble can be changed if within the tuple).
    - allows duplicate members.
-  Set
      - unordered
      - unindexed
      - no duplicate members.
-  Dictionary
       - unordered
       - changeable
       - indexed.
       - no duplicate members. If here are duplicates, then it will always take the first one



## Tuples

```python
a = [5] # will be a list as it uses BRACKETS
e = (5,) # will be a tuple as it uses PARENTHESES

type(a), type(e)
```

```python
b = (5)  # will not be a tuple - because it is only one item and no comma
c = (5,) # will be a tuple, as it contains the comma

type(b), type(c)
```

```python
t = (a, 'aba', 12, 1.20)
t1 = a, 'aba', 12, 1.20  # no need for parentheses, but it is a best practice

type(t), type(t1)
```

```python
print( (*t1, 6, 5) ) # asterisks "unpacks" the tuple, which has the same effect as the following:

print( t1 + (6, 5) ) 

print(  (t1, 6, 5) ) # but this is not unpacked 
```

```python
print( t[1]          )
print( t[0]          ) 
print( t[0:1][0]     ) 
print( type(t[3])    ) 
print( t.index(12)   ) 
print( t.count(12)   ) 
```

```python
t[0] = 5 # expected error, tuples are unchangeable
```

## Sets

```python
s = set(['u', 'd', 'u'])
s
```

```python
set1 = {'1', '2', 'a', 'b'}
set1[1]  # expected error
```

```python
set1.add('hotel')
set1
```

```python
set1.add(['hotel', 'villa'] )
# returns unhashable error. Only immutable objects (so not lists), can have a hash value
# hash value acts as a "fingerprin" for an object, that should never change
```

```python
set1.update(["orange", "mango", "grapes"])  # to add more than one item
set1
```

```python
set1.remove('120')  
# if the item does not exist, this will raise an error, thats where discard comes in
```

```python
# thats why its better to use discard
set1.discard('120')
```

```python
set1.clear()
set1
```

```python
set2 = s + set1  # does not work
```

```python
set2 = s.union(set(['b', 'u']), s)
set2
```

```python
s.intersection(set2)
```

```python
set2.difference(s)  # items in set2 but not in set s
```

```python
u = {'v', 'u'}
u.symmetric_difference(s)  # items in one or the other but not in both
```
