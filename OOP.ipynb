```python
class FinInst():
    def __init__(self, symbl, prc):
        self.symbol = symbl
        self.__price = prc  # private

    def get_price(self):
        return self.__price

b = FinInst('aapl', 10)

b.get_price() 
```

```python
b.__price   # is a private attribute, so this will result in an error
```

```python
b._FinInst__price  # If the class name is prepended with a single leading underscore, direct access and manipulation are still possible.
```

```python
b._FinInst__price = 500
b._FinInst__price
```

## Inheritance
- when class inherits methods of another class, we can even have a "multiple inheritance" - class inheriting from several classes

```python
# Inheritance
class BondInstrument(FinInst):  # inherits FinInst
    def set_price(self, price):
        self._FinInst__price = price # !!!! price is attribute of the FinInst not FinInstrumet object, thats why it has to have the prefix
```

```python
inst = BondInstrument('aapl', '800')
inst.symbol
```

```python
inst.get_price() # function not defined for FinInstrument but only for FinInst
```

```python
inst.set_price(9000)
inst.get_price()
```

### super() method
built-in super() function allows an overriding method to call the original method in the parent class.
it can be used, when there is a child class that would have the same funnction as the parent class - but we want to add a bit extra to the one in the child class

```python
class Parent():
    def welcome(self):
        print("Welcome everyone")
        print("We hope you have an amazing experience")

class Child(Parent):
    def welcome(self):
        super().welcome() # this will call the function from the parent class
        print("but as opposed to the parent class, this also says \"Goodbye\" ")
```

```python
a = Parent()
a.welcome()
```

```python
a = Child()
a.welcome()
```

```python
issubclass(Child, Parent) # we can check if a child is a subclass of a parent
```

## Agregation

```python
class PortPos():
    def __init__(self, FinInstrument, size):
        self.size = size
        self.position = FinInstrument  # uses a different class

    def value_portfolio(self):
        return self.size * self.position.get_price() #can aggregate across different classes
```

### Class Methods

- class methods are associated with a class rather than with individual objects - you need to use @before the name of the method, and cls needs to be the first parameter
- using the class method, I can assign the class to an object and call a method at the same time
- not frequently used Python feature

```python
class ExampleClass:
    def exampleRegularMethod(self):
        print('regular method')

    @classmethod
    def exampleClassMethod(cls):
        print('class method')

ExampleClass.exampleClassMethod()

obj = ExampleClass()
obj.exampleClassMethod() 
```
