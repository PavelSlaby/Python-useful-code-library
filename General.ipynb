## Identity and Hash

```python
a = 5
b = 5
b is a
```

```python
x = 5
y = x
id(x), id(y)  # identity function, address of the variable in memory, #memory address of the object is same for x and y
```

```python
z = 5
id(z)
```

```python
a = (4, 5)
b = (4, 5)
hash(a), hash(b) # hash is a unique fingerprint
```

```python
a = [4] 
hash(a) # throws an error because lists are not hasable - only objects that can not change are hashable
```
