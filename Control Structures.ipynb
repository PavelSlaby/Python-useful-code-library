# Control Structures

### For Cycle

```python
a = [1, 2, 3, 4, 5, 6, 7, 8]

for neco in a:
    print(neco, end = ' ')
```

```python
for i in a[0:4]:
    print(i ** 2)
```

```python
r = range(0, 8, 1)

for i in range(10):
    print(i, end=" ")
```

```python
# Break
for i in "Hello World":
    print(i)
    if i == "W": break
```

```python
# Continue
for i in "Hello World":
    if i == "W": continue  # continues with the next item
    print(i)
```

```python
# you can end for with else
for i in range(0, 30, 3):
    print(i)
else:
    print('finito')
```

```python
# While Cycle
```

```python
total = 0
while total < 100:
    total += 1
    print(total, end=" - ")
```

```python
i = 0
while i < 6:
    print(i)
    i += 1
    if i == 3: break
    if i == 2:
        continue
    print('2a')
else:
    print('finito')
```

### if statement

```python
if 5 > 4: print(1)
```

```python
print(True) if 5 > 4 else print(False)
print(True) if 5 > 6 else print(False) if 4 < 2 else print('c')
```

```python
if 5 > 3: pass  # if statements cannot be empty, put in the pass statement to avoid getting an error.
```

```python
for i in range(1, 20):
    if i % 2 == 0:  # % is for modulo
        print("i is even number")
    elif i % 3 == 0:
        print("i is dividable by 3")
    else:
        print("number is odd")
```
