# TIME FUNCTIONS

Key classes:
* datetime.datetime – full timestamp (date + time)
* datetime.date – date only
* datetime.time – time only
* datetime.timedelta - time diff


```python
import datetime as dt
from datetime import date, datetime, timedelta
```

```python
dt.datetime(2022, 1, 1)
```

```python
dt.date(2022, 1, 1)
```

```python
dt.time(23, 59, 59)
```

```python
dt.datetime.today()
```

```python
str(dt.datetime.today())
```

```python
dt.datetime.today().strftime('%Y-%m-%d'), type(dt.datetime.today().strftime('%Y-%m-%d'))
```

```python
dtnow = dt.datetime.today()
dtnow.month
```

```python
dtnow.year
```

```python
dtnow.day
```

```python
dtnow.strftime('%A'), dtnow.strftime('%B')
```

```python
dtnow + timedelta(days=365)
```

```python
dt.datetime.today().strftime('%m/%d/%Y %H:%M'), dt.datetime.today().strftime('%Y-%m-%d %H:%M')
```
