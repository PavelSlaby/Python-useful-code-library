# Functions

```python
def myfunct(*args):  # if i do not know how many parameters will be passed
    print(args)

a = 6

# Arbitrary Arguments are often shortened to *args
myfunct(4, 5, a)
```

```python
# type hints
def random_func(x:int) -> str:
    if x < 0: return 'negative'
    if x > 0: return 'positive'

random_func(True)  #  the type hint can be violated and the code still runs, the IDE ignores type hints
```

```python
# keyword arguments = **kwargs
def my_function(**kwargs):
    print("His last name is " + kwargs["lname"])

my_function(fname="Tobias", lname="Refsnes")
```

```python
def nasob2(x, y, z=None, a=False):
    if z == None:
        return x * y
    elif a:
        return x / y / z
    else:
        return x * y * z

a = nasob2
a(4, 5, 6, 1 == 1)
```

```python
def power(x):
    return x ** 2

power(2)
```

```python
def addition(n):
    return n + n

numbers = (1, 2, 3, 4)
k = map(addition, numbers)
list(k)
```

```python
def f():
    return True

if f():
    print("truth")
```

```python
# Lambda -  A lambda function can take any number of arguments, but can only have one expression.
x = lambda a: a + 10
x(10)
```

```python
x = lambda a, c: a ** c if 5 < 3 else c
x(6, 2)
```

```python
def myfunc(n):
    return lambda a: a * n

mydoubler = myfunc(2)
mytripler = myfunc(3)

mydoubler(11), mytripler(11)
```

```python
# Higher order function
def math_operation(func, *args):
    print(func(args))

math_operation(min, 1, 2, 3)
math_operation(max, 1, 2, 3)
```

```python
list(map(lambda n:min(n), [[2, 1], [3, 4]]))
```

```python
# Filter
ages = [5, 12, 17, 18, 24, 32]

def myFunc(x):
    if x < 18:
        return False
    else:
        return True


list(filter(myFunc, ages))
```

```python
# using "global" variable - valid outside function
def myfunc():
    global x
    x = "fantastic"
    print(x)

def myfunc2():
    x = "fantastic"
    print(x)
```

```python
x = 'not-fantastic'
x
```

```python
myfunc()
x  # x was rewritten by the previous use of the myfunct function
```

```python
x = 'not-fantastic'
myfunc2()
x  # this time it was not rewritten
```
